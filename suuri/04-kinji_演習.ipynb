### 🔹 ノート 1 math は数学の道具箱（スライド 5-1）

1 行目の `import math` で、数学の関数が入ったモジュール（道具箱）`math` を読み込みます。`math.sqrt` は「math の中の sqrt（平方根）」です。三角関数の角は、度ではなくラジアンで渡します（30° は `math.pi / 6`）。

### 🔹 ノート 2 関数を作って傾きを出す（スライド 5-2）

`def f(x):` で関数 f を作り、次の行の `return` で x² を返します（2 行目は空白 4 つで字下げします）。x ＝ 3 での差の商を、h を小さくしながら出します（h のリスト hs の 1e-08 は 10⁻⁸ の書き方です）。

```python
def f(x):
    return x ** 2

a = 3
hs = [1, 0.1, 0.01, 0.001, 1e-08, 1e-12, 1e-16]
for h in hs:
    print(h, (f(a + h) - f(a)) / h)
```

**数を替えてみる**

2 行目の `return x ** 2` を `return x ** 3` に替えて実行すると、4 行目の出力は `0.001 27.009000999996147` で、x³ の x ＝ 3 での傾き 27 に近い値です。

### 🔹 ノート 3 SymPy は文字のまま計算する（スライド 5-3）

SymPy は、x を文字のまま扱う道具箱です。`sp.diff(f, x)` は f を x で微分した式を返し、`.subs(x, 3)` で式の x に 3 を入れます。ノート 2 の差の商と見比べます。

```python
import sympy as sp
x = sp.symbols("x")       # 数の入っていない文字 x
f = x ** 2
df = sp.diff(f, x)        # f を x で微分した式
print(df)
print(df.subs(x, 3))      # 式の x に 3 を入れた値
```

**数を替えてみる**

3 行目の `f = x ** 2` を `f = x ** 3` に替えて実行すると、2 行目の出力は `27` です。

### 🔹 ノート 4 SymPy でテイラー展開を出す（スライド 5-4）

`sp.series(f, x, 0, n)` は、f を x ＝ 0 のまわりで、x の n 乗より手前まで読んだ式を返します（3 次までなら n ＝ 4）。`O(x**4)` は「x の 4 乗から先を打ち切った」印で、`removeO()` で外すと数を入れられます。

```python
import sympy as sp
x = sp.symbols("x")
for n in [2, 3, 4, 5]:
    print(sp.series(sp.exp(x), x, 0, n))
# O を外した式（removeO）
p = sp.series(sp.exp(x), x, 0, 4).removeO()
print(p.subs(x, 1), sp.N(p.subs(x, 1)))
```

**数を替えてみる**

6 行目の `sp.series(sp.exp(x), x, 0, 4)` を `sp.series(sp.exp(x), x, 0, 6)` に替えて実行すると、最後の行は `163/60 2.71666666666667` で、e ＝ 2.71828… に近づきます。

### 🔹 ノート 5 計算機でそのまま足すと壊れる和（スライド 5-5）

$e^x$ のテイラーの和 1 ＋ x ＋ x²/2! ＋ … に x ＝ −20 を入れて 100 項足し、`math.exp(-20)` と比べます。total はテイラーの和をそのまま足した値、big は $e^{20}$ の和です。最後の行は、1 を big で割る書き方です。

```python
import math
x = -20
total = 0
for n in range(100):
    total = total + x ** n / math.factorial(n)
print(total)
print(math.exp(x))

big = 0
for n in range(100):
    big = big + 20 ** n / math.factorial(n)
print(1 / big)
```

**数を替えてみる**

2 行目の `x = -20` を `x = -5` に替えて実行すると、1 行目の出力は `0.006737946999086907` で、2 行目の `math.exp(x)` とほぼ同じです。

### 🔹 ノート 7 剰余で外れを見積もる（スライド 6-4）

$e = e^1$ を、$e^x$ の 1 ＋ x ＋ x²/2! ＋ … の n 次まで足した値と、本当の外れ e − 近似、剰余の上限 3/(n＋1)! を並べます。外れは、どの n でも上限より小さくなります。

```python
import math
for n in [1, 2, 3, 4, 5, 6]:
    approx = sum(1 / math.factorial(k) for k in range(n + 1))    # e を n 次まで足した値
    bound = 3 / math.factorial(n + 1)                            # 剰余の上限（e < 3 を使う）
    print(n, approx, math.e - approx, bound)
```

**数を替えてみる**

2 行目の `[1, 2, 3, 4, 5, 6]` を `[1, 2, 3, 4, 5, 6, 8]` に替えて実行すると、最後の行は `8 2.71827876984127 3.0586177750535626e-06 8.267195767195768e-06` です。
