### 🔹 ノート 1 登るほど減り方がゆるくなる気圧（スライド 2-1）

2-1 の模型で、高さ 0・1,000・3,776 m の気圧と、傾き（100 m あたり）を出します。傾きは、今の気圧を H で割ってマイナスを付けた値です。

```python
import math
P0 = 1013.25    # 海面の標準の気圧（hPa）
H = 8434.5      # 気圧が 1/e になる高さ（m）。気温を 15 ℃ で一定と置いた模型
for h in [0, 1000, 3776]:
    P = P0 * math.exp(-h / H)
    print("高さ", h, "m: 気圧", round(P, 2), "hPa・傾き 100 m あたり", round(-P / H * 100, 2), "hPa")
```

**数を替えてみる**

4 行目の `[0, 1000, 3776]` を `[0, 1000, 5000]` に替えて実行すると、3 行目の出力は `高さ 5000 m: 気圧 560.1 hPa・傾き 100 m あたり -6.64 hPa` です。高い所ほど、減り方（傾きの大きさ）が小さくなります。

### 🔹 ノート 2 計算機の小数は対数の目盛りで並ぶ（スライド 5-1）

Python の `math.ulp(x)` は、x と、計算機が表せる x のすぐ上の小数との差（刻み）を返します。1・1,024・10³⁰⁰ の近くの刻みと、刻み ÷ 数を出します。

```python
import math
for x in [1.0, 1024.0, 1e300]:
    print(x, "の近くの刻み:", math.ulp(x), " 刻み ÷ 数:", math.ulp(x) / x)
```

**数を替えてみる**

2 行目の `[1.0, 1024.0, 1e300]` を `[1.0, 2048.0, 1e300]` に替えて実行すると、2 行目の出力は `2048.0 の近くの刻み: 4.547473508864641e-13  刻み ÷ 数: 2.220446049250313e-16` です。数が 2 倍になると、刻みも 2 倍になります。

### 🔹 ノート 3 eˣ の傾きは eˣ（スライド 6-1）

底 a の aˣ について、x ＝ 0 での差の商 $\frac{a^h-1}{h}$ を h ＝ 0.0001 で出し、ln a と並べます。a ＝ e のとき、差の商は 1 に近く、ln a は 1 です。

```python
import math
h = 0.0001
for a in [2, math.e, 3]:
    print("a =", round(a, 4), "差の商", round((a ** h - 1) / h, 4), " ln a", round(math.log(a), 4))
```

### 🔹 ノート 4 逆向きに読むと傾きは逆数（スライド 6-3）

SymPy の `sp.diff` で、ln x と eˣ の傾きの式を出します。SymPy は ln x を `log(x)` と書きます。最後の 2 行で、ln x の傾きが eˣ の傾きの逆数になっていることを確かめます。

```python
import sympy as sp
x = sp.symbols("x")
print(sp.diff(sp.log(x), x))    # ln x の傾きの式（SymPy は ln を log と書きます）
print(sp.diff(sp.exp(x), x))    # eˣ の傾きの式
a = 2
print(float(sp.exp(a)), float(1 / sp.exp(a)))    # x ＝ a での eˣ の傾きと、その逆数
print(float(sp.diff(sp.log(x), x).subs(x, sp.exp(a))))    # ln x の、x ＝ eᵃ での傾き
```

**数を替えてみる**

5 行目を `a = 3` に替えて実行すると、最後の行は `0.049787068367863944` で、その上の行の 2 つ目の数と同じです。

### 🔹 ノート 5 arctan の傾きは 1/(1＋x²)（スライド 6-5）

SymPy で arctan x と arcsin x の傾きの式を出します。SymPy では `sp.atan`・`sp.asin` と書きます。

```python
import sympy as sp
x = sp.symbols("x")
print(sp.diff(sp.atan(x), x))    # arctan x の傾きの式
print(sp.diff(sp.asin(x), x))    # arcsin x の傾きの式
```

**数を替えてみる**

4 行目の `sp.asin(x)` を `sp.acos(x)` に替えて実行すると、2 行目の出力は `-1/sqrt(1 - x**2)` で、arcsin の傾きにマイナスが付いた形です。
