### 🔹 ノート P2 一般項を並べて、和の公式と比べる（スライド 5-2）

初項 a・公差 d の等差数列の第 n 項 `a + d*(n-1)` を、n が 1 から N までの順に、角かっこの中の 1 行で並べます（リストの内包表記）。`sum` で足した値と、和の公式 N × (初項 ＋ 最後の項) ÷ 2 の値を並べます。公式の割り算は `//` で書いています。`/` で割ると、答えが `679.0` のように小数の形で出ます。

```python
a, d, N = 3, 7, 14
seq = [a + d*(n-1) for n in range(1, N+1)]
print(seq[0], seq[-1], len(seq))
print(sum(seq))
print(N * (seq[0] + seq[-1]) // 2)
```

**数を替えてみる**

`a, d, N = 3, 7, 10` に替えると、出力の 1 行目は `3 66 10` です。sum で足した値と公式の値は、ここでも同じです。

### 🔹 ノート P3 np.diff で差を取り、np.cumsum で足し上げる（スライド 5-3）

`hi` は、東京の 2 月 1 日から 7 日までの日最高気温（℃。2-1 の値）です。`np.cumsum` は先頭から足し上げた値を並べ、`np.diff` は隣どうしの差（後ろ − 前）を並べます。`prepend=0` を付けると、先頭に 0 を置いてから差を取ります。後半は、1 から 8 までの 2 乗の並び `sq` の差を、1 回と 2 回取ります。

```python
import numpy as np
hi = np.array([11.1, 12.0, 10.8, 11.4,
               13.8, 16.1, 6.9])
print(np.cumsum(hi))
print(np.diff(hi))
print(np.diff(np.cumsum(hi)))
print(np.diff(np.cumsum(hi), prepend=0))
sq = np.arange(1, 9)**2
print(sq)
print(np.diff(sq))
print(np.diff(sq, n=2))
```

**数を替えてみる**

最後の行を `np.diff(sq, n=3)` に替えると、出力の最後の行は `[0 0 0 0 0]` です。2 回目の差がどこも 2 なので、3 回目の差には 0 が並びます。

### 🔹 ノート P4 SymPy の summation と rsolve（スライド 5-4）

`summation(式, (k, 1, n))` は、k を 1 から n まで動かした和を、式のまま出します。上の端の n の項も和に入ります。`factor` は式を因数分解します。`rsolve` は、漸化式（前の項から次の項を作る式）から一般項を出します。ここで渡したのは 3-2 の上乗せの式 a(c ＋ 1) ＝ 2a(c) ＋ 1 と、a(0) ＝ 0 です。

```python
from sympy import symbols, summation, factor
from sympy import Function, rsolve
k, n, c = symbols('k n c')
print(summation(k, (k, 1, n)))
print(factor(summation(k, (k, 1, n))))
print(summation(k**2, (k, 1, n)))
print(factor(summation(k**2, (k, 1, n))))
print(summation(2**k, (k, 0, n - 1)))
a = Function('a')
print(rsolve(a(c + 1) - 2*a(c) - 1, a(c), {a(0): 0}))
```

**数を替えてみる**

4 つ目の print の中の `k**2` を `k**3` に替えると、出力の 4 行目は `n**2*(n + 1)**2/4` です。出力の 2 行目の n*(n + 1)/2 を 2 乗した式です。

### 🔹 ノート P5 0.1 を 10 回足すと 1 になるか（スライド 5-5）

`xs` は 0.1 を N 個並べたリストです。for で 1 つずつ足した値・組み込みの `sum` で足した値・`math.fsum` で足した値を、正しい値 N ÷ 10 との差と一緒に、1 行ずつ出します。

```python
import math
N = 10
xs = [0.1] * N
total = 0.0
for x in xs:
    total += x
print(total, total - N / 10)
print(sum(xs), sum(xs) - N / 10)
print(math.fsum(xs), math.fsum(xs) - N / 10)
```

**数を替えてみる**

`N = 20` に替えると、出力の 1 行目は `2.0000000000000004 4.440892098500626e-16` です。for で足した値は、今度は 2 より少し大きくなります。

## 等比数列の和と、差の調べ方（🔹）

### 🔹 ノート P6 比が一定なら等比数列（スライド 6-2）

初項 a1・公比 r の等比数列の、第 n 項までの和を記号のまま出します。`simplify` で整えると、r ＝ 1 の場合とそれ以外の場合に分けた式（`Piecewise`）が返ります。出力の 2 行目は、a1 ＝ 1、r ＝ 2、n ＝ 5 を入れた値です。

```python
from sympy import symbols, summation, simplify
a1, r = symbols('a1 r')
k, n = symbols('k n', integer=True, positive=True)
S = summation(a1 * r**(k - 1), (k, 1, n))
print(simplify(S))
print(S.subs({a1: 1, r: 2, n: 5}))
```

**数を替えてみる**

2 つ目の print の `r: 2` を `r: 1` に替えると、出力の 2 行目は `5` です。r ＝ 1 の場合の式 a1*n の値です。

### 🔹 ノート P7 1 回目の差・2 回目の差・比で並びを調べる（スライド 6-4）

6-4 の例の 3 つの並びで、1 回目の差 `np.diff(x)`・2 回目の差 `np.diff(x, n=2)`・隣どうしの比 `x[1:] / x[:-1]` を順に取ります。3 行ずつ、並びの順に出ます。

```python
import numpy as np
seqs = [
    [200, 250, 300, 350, 400, 450],
    [1, 4, 9, 16, 25, 36],
    [1, 2, 4, 8, 16, 32],
]
for s in seqs:
    x = np.array(s)
    print(np.diff(x))
    print(np.diff(x, n=2))
    print(x[1:] / x[:-1])
```

**数を替えてみる**

2 つ目の並びを `[1, 3, 6, 10, 15, 21]` に替えると、出力の 5 行目は `[1 1 1 1]` です。2 回目の差が 1 で一定になります。
