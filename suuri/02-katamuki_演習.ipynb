### 🔹 ノート 1 1 日に約 2 分ずつ短くなる昼（スライド 2-1）

国立天文台 暦計算室の日の出入りの表（2026 年・東京）から出した昼の長さ（分）です。差を取る日数で割ると、1 日あたりの変わり方になります。

```python
m_29 = 714   # 9 月 29 日の昼の長さ（分）
m_06 = 700   # 10 月 6 日
m_13 = 684   # 10 月 13 日
print("次の 7 日の差", m_13 - m_06, "分")
print("次の 7 日の差を 7 で割ると", round((m_13 - m_06) / 7, 1), "分")
print("前後 7 日の差を 14 で割ると", round((m_13 - m_29) / 14, 1), "分")
```

**数を替えてみる**

### 🔹 ノート 2 差を取る幅を縮めすぎると（スライド 5-1）

x ＝ 1 で、差の商 $\frac{\sin(1+h)-\sin 1}{h}$ を h ＝ 10⁻¹ から 10⁻¹⁶ まで出します。7 桁に丸めて、最後の行の cos 1 と見比べます。

```python
import math
for k in range(1, 17):
    h = 10 ** -k
    print("h = 10 の", -k, "乗:", round((math.sin(1 + h) - math.sin(1)) / h, 7))
print("cos 1 =", round(math.cos(1), 7))
```

**数を替えてみる**

4 行目の `(math.sin(1 + h) - math.sin(1)) / h` を `(math.sin(1 + h) - math.sin(1 - h)) / (2 * h)`（前後に取る差）に替えて実行すると、10 の −16 乗の行は `h = 10 の -16 乗: 0.5551115` で、0 になりません。

### 🔹 ノート 3 差の商の行き先が傾き（スライド 6-1）

$f(x)=x^3$ の a ＝ 2 での差の商 $\frac{(a+h)^3-a^3}{h}$ を、h を 1 から 0.001 まで縮めて出します。最後の行は傾き $3a^2$ です。

```python
a = 2     # 傾きを出す点
for h in [1, 0.1, 0.01, 0.001]:
    print("h =", h, "差の商", round(((a + h) ** 3 - a ** 3) / h, 6))
print("傾き 3a² =", 3 * a ** 2)
```

**数を替えてみる**

1 行目を `a = 3` に替えて実行すると、h ＝ 0.001 の行は `h = 0.001 差の商 27.009001` で、最後の行の傾き 27 に近づきます。

### 🔹 ノート 4 xⁿ の傾きは n xⁿ⁻¹（スライド 6-2）

SymPy の `sp.diff` で、$x^3$・$\frac{1}{x}$・$\sqrt{x}$ の傾きの式を出します。$\sqrt{x}$ は `sp.sqrt(x)` と書きます。

```python
import sympy as sp
x = sp.symbols("x")
print(sp.diff(x ** 3, x))
print(sp.diff(1 / x, x))
print(sp.diff(sp.sqrt(x), x))
```

**数を替えてみる**

3 行目の `x ** 3` を `x ** 10` に替えて実行すると、1 行目の出力は `10*x**9` で、n xⁿ⁻¹ の形のままです。

### 🔹 ノート 5 sin と cos の傾き（スライド 6-5）

SymPy で sin と cos の傾きの式を出し、x ＝ 1 での前後の差の商と cos 1 を並べます。

```python
import math
import sympy as sp
x = sp.symbols("x")
print(sp.diff(sp.sin(x), x))
print(sp.diff(sp.cos(x), x))
h = 0.00001
print(round((math.sin(1 + h) - math.sin(1 - h)) / (2 * h), 7), round(math.cos(1), 7))
```

**数を替えてみる**

最後の行の `math.sin(1 + h) - math.sin(1 - h)` を `math.cos(1 + h) - math.cos(1 - h)` に、`round(math.cos(1), 7)` を `round(-math.sin(1), 7)` に替えて実行すると、最後の行は `-0.841471 -0.841471` です。cos の傾きは −sin です。
