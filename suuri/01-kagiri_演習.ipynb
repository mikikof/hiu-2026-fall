### 🔹 ノート 1 冷めていくコーヒー（スライド 2-1）

室温との差が、1 分ごとに 0.95 倍に縮みます。`**` は「何乗」、`*` は「かけ算」です。

```python
room = 25    # 室温（℃）
start = 90   # 初めの温度（℃）
r = 0.95     # 室温との差が、1 分で何倍になるか
for t in [0, 10, 20, 30, 60, 120]:    # t に 0, 10, 20, … を順に入れて、下の行を繰り返します
    print(t, "分後", round(room + (start - room) * r ** t, 1), "℃")
```

**数を替えてみる**

3 行目を `r = 0.9` に替えて実行すると、30 分後の行は 39.0 ℃ から 27.8 ℃ に下がります。r が小さいほど、早く室温に近づきます。

### 🔹 ノート 2 当たり 1% を 100 回引く（スライド 3-1）

1 回ごとに確率 p（ここでは 1%）で当たり、前の結果によらないくじを n 回引きます。1 回も当たらない確率は、外れる確率 $1-p$ を n 回かけた $(1-p)^n$ です。

```python
p = 0.01   # 1 回で当たる確率
for n in [10, 50, 100, 200, 500]:
    print(n, "回で 1 回も当たらない確率", round((1 - p) ** n * 100, 1), "%")
```

**数を替えてみる**

1 行目を `p = 0.002`（500 回に 1 回）に替えて実行すると、500 回の行は `500 回で 1 回も当たらない確率 36.8 %` になります。この 36.8% は、1/e（0.3678…）を百分率にして丸めた値と同じです（スライド 6-5）。

### 🔹 ノート 3 攻略 AI が先の報酬を割り引いて足す（スライド 4-1）

足す項の数 k を増やすと、$1+r+r^2+\cdots$ の和は $\frac{1}{1-r}$ に近づきます（r ＝ 0.9 なら 10）。

```python
r = 0.9    # 割引の率
for k in [1, 10, 50, 100, 300]:
    total = sum(r ** i for i in range(k))    # 1 + r + r**2 + … を k 項まで足します
    print(k, "項まで足すと", round(total, 4))
```

**数を替えてみる**

1 行目を `r = 0.99` に替えて実行すると、300 項の行は `300 項まで足すと 95.0959` で、行き先の 100 にまだ届きません。

### 🔹 ノート 4 計算機で確かめると行き先がずれる（スライド 5-1）

n を 10 倍ずつ大きくして、$\left(1+\frac{1}{n}\right)^n$ を計算します。スライド 5-1 の計算機の値と見比べます（最後の 1 桁が、計算機によってはスライドと違うことがあります）。

```python
import math
print("e =", math.e)
for k in range(1, 18):
    n = 10 ** k
    print("n = 10 の", k, "乗:", (1 + 1/n) ** n)
```

**数を替えてみる**

5 行目の `(1 + 1/n) ** n` を `1 + 1/n` に替えて実行すると、16 乗の行は `n = 10 の 16 乗: 1.0` です。かっこの中が、すでに 1 になっています。

### 🔹 ノート 5 0/0 は約分してから（スライド 6-4）

$$\lim_{x\to 2}\frac{x^{2}-4}{x-2}$$

x を 2 の両側から近づけて、式の値が 4 に寄ることを確かめます。

```python
for x in [1.9, 1.99, 1.999, 2.001, 2.01, 2.1]:
    print(x, round((x**2 - 4) / (x - 2), 6))
```
