## 1. 比べた答え

> **対応スライド**: 「比べた答えは True か False」「'18' と 18 を比べると」

比べた答えは、True（正しい）か False（正しくない）です。この2つの値の型を bool と呼びます。

🔹 実行します。

````python
print(75 >= 60)
print(type(75 >= 60))
print('18' == 18)
````

## 2. if と else

> **対応スライド**: 「if 文の形」「そうでないときの行」

`if 条件:` の次の行を字下げすると、条件が True のときだけ、その行が動きます。
`else:` の下に、条件が False のときに動く行を書きます。

🔹 `40` と打って `Enter` を押します。

````python
fun = int(input('今週、課題にかけた分: '))
if fun >= 60:
    print('目安の60分に届きました')
else:
    print(f'あと {60 - fun} 分で目安です')
print('記録しました')
````

## 4. elif

> **対応スライド**: 「3つ以上に分ける」

`elif` で条件を足すと、3つ以上に分けられます。上から順に調べ、最初に True になった1つだけが動きます。

🔹 `72` と打って `Enter` を押します。

````python
ten = int(input('小テストの点: '))
if ten >= 80:
    print('よくできました')
elif ten >= 60:
    print('合格')
else:
    print('再テスト')
````

## 5. 境い目

> **対応スライド**: 「> と >= の違いは境い目だけ」

`>` と `>=` で答えが変わるのは、境い目の値のときだけです。

🔹 `18` と打って `Enter` を押します。

````python
nenrei = int(input('年齢: '))
if nenrei > 18:
    print('成人です')
else:
    print('未成年です')
````

確かめる値は、境い目とその両隣の3つです。

## 6. and・or・not

> **対応スライド**: 「条件を組み合わせる」

and は両方が True のとき、or はどちらかが True のときに True です。not は True と False を入れ替えます。

🔹 実行します。

````python
ten = 72
print(ten >= 0 and ten <= 100)
youbi = '土'
print(youbi == '土' or youbi == '日')
ten = 72
print(not ten >= 60)
````

### 練習4（修正）― 気温のメッセージ

✏️ 気温（度・整数）を打つと、25 以上なら「暑い日です」、15 以上なら「過ごしやすい日です」、それより低ければ「寒い日です」と出るように直します。

考えどころ: エラーは出ません。3つのメッセージが全部出るか、打つ数を自分で選んで確かめます。スライド「並べる順を逆にすると」が手がかりです。

````python
kion = int(input('今日の気温: '))
if kion >= 15:
    print('過ごしやすい日です')
elif kion >= 25:
    print('暑い日です')
else:
    print('寒い日です')
````
