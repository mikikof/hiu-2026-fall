## 1. 演算の答えの型

> **対応スライド**: 「演算の答えの型」

演算子は、答えの型まで決まっています。`/` の答えは、割り切れても float です。

🔹 実行します。

````python
print(6 / 3, type(6 / 3))
print(7 // 2, type(7 // 2))
print(90 + 0.5, type(90 + 0.5))
````

## 2. 文字列に使える演算

> **対応スライド**: 「文字列に使える演算」「同じ順位の計算」

`+` は文字列どうし、`*` は文字列と int のときだけ動きます。
同じ順位の演算子が並ぶと、左から計算します。`**` だけは右から計算します。

🔹 実行します。

````python
print('90' * 2)
print(100 / 10 * 2)
print(2 ** 3 ** 2)
````

## 4. 打った数字は文字列

> **対応スライド**: 「打った数字は文字列」

数字を打っても、input が返すのは文字列です。

🔹 `1350` と `60` を順に打ちます。

````python
a = input('1つめの数: ')
b = input('2つめの数: ')
print(a + b)
print(type(a))
````

## 5. 電卓を組む

> **対応スライド**: 「型を変える」「電卓を組む」

受け取った直後に int() で数に変えると、電卓になります。

🔹 `1350` と `60` を順に打ちます。

````python
a = int(input('1つめの数: '))
b = int(input('2つめの数: '))
print(a + b)
print(a / b)
print(a // b, a % b)
````
