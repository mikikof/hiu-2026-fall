### 練習8（修正）― 小数の長さから面積を出す

> **対応スライド**: 「型を変える」「エラーと途中の値の見方」

✏️ 長方形の縦と横の長さを cm で受け取り、面積を小数点以下1桁で出すプログラムに直してください。
長さには小数も打てるようにします。長方形の面積は、縦と横の長さを掛けた値です。
聞く文と最後の `print` はそのままにし、受け取った値を使って計算してください。

まず縦に `2.5` を打ち、横を聞かれたら `4` を打ちます。止まった場合は、エラーのいちばん下の行を読んで直し、セルをもう一度実行します。

**確かめ方**: 聞く文の行のあとに、次の1行が出れば、直っています。エラーが消えたあとも、面積を見比べます。

```
面積 10.0 cm²
```

````python
# 小数の長さを受け取り、長方形の面積を出す
tate = int(input('縦の長さを cm で: '))
yoko = float(input('横の長さを cm で: '))
menseki = tate + yoko
print(f'面積 {menseki:.1f} cm²')
````

### 練習9（書く）― 読書の進み具合を出す

> **対応スライド**: 「電卓を組む」「計算の結果を文に埋め込む」

✏️ 本の全体のページ数と、読んだページ数をこの順に `input` で受け取り、残りのページ数と進み具合を出すプログラムを書いてください。

- 受け取った値は整数にします。全体のページ数には正の整数を使い、読んだページ数は0以上、全体以下とします。
- 残りのページ数は、全体から読んだページ数を引いて求めます。
- 進み具合は、読んだページ数が全体の何％かを、小数点以下1桁で出します。割合は「読んだページ数 ÷ 全体のページ数 × 100」です。
- 見本の数を直接 `print` に書かず、受け取った値から計算します。聞く文は、出力の項目名と違う言い方にします。

ヒント: 整数で受け取っても、割り算の途中では小数が必要です。表示する桁は f 文字列の `:.1f` で決めます。

**確かめ方**: 全体に `96`、読んだページ数に `35` を打つと、聞く文の行のあとに次の2行が出ます。`%` の前には空白を入れません。

```
残り 61 ページ
進み具合 36.5%
```
