## 2. 同じ結果でも、手順は1つではない

> **対応スライド**: 「同じ結果でも、手順は1つではない」

1 から 10 までの合計を、2つの手順で出します。**この2つは、書けるようになる必要はありません。**
変数は第4回、`for` は第6回で扱います。

🔹 実行して、答えが同じになることを確かめてください。

````python
# 手順A 1つずつ足す（足し算 10 回）
total = 0
for i in range(1, 11):
    total = total + i
print(total)
````

🔹 もう1つの手順です。足し算は1回だけで、出るものは同じになります。

````python
# 手順B まとめて計算する（足し算 1 回）
print((1 + 10) * 10 // 2)
````
