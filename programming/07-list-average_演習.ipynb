## 1. リストと番号

> **対応スライド**: 「リストの書き方」「番号で取り出す値」「最後の番号と個数」

リストは、複数の値を `[ ]` で囲み、カンマで区切って作ります。
`len` は値の個数を返します。番号は0から始まります。

🔹 学食のメニューの値段をまとめます。値段の単位は円です。

````python
prices = [360, 420, 480]
print(prices)
print(len(prices))
print(type(prices))
print(prices[0])
print(prices[2])
````

## 2. for で回す・合計と平均

> **対応スライド**: 「リストを使う繰り返し」「合計と平均」

`for` にリストを渡すと、先頭から順に値を取り出します。

🔹 メニューの値段を1つずつ出し、合計も求めます。

````python
prices = [360, 420, 480]
total = 0
for price in prices:
    print(f'{price} 円')
    total = total + price
print(f'合計 {total} 円')
````

`sum` は数値のリストの合計を返します。合計を個数で割ると平均です。
`/` の結果は、割り切れる場合も小数になります。

🔹 同じ値段から合計と平均を出します。

````python
prices = [360, 420, 480]
print(f'合計 {sum(prices)} 円')
print(f'平均 {sum(prices) / len(prices)} 円')
````

## 3. append

> **対応スライド**: 「空のリストへの追加」「打ち込んだ点数の平均」

`[]` は空のリストです。`append` で、末尾に値を1つ足します。

## 5. 内包表記

> **対応スライド**: 「1行で作る新しいリスト」

内包表記では、取り出した値をどう変えるかを先に書きます。
次の2つのセルは、曲の長さを分から秒に換算したリストを作ります。

🔹 `for` と `append` で作ります。

````python
minutes = [3, 4, 6]
seconds = []
for minute in minutes:
    seconds.append(minute * 60)
print(seconds)
````

🔹 内包表記で作ります。先ほどの出力と見比べます。

````python
minutes = [3, 4, 6]
seconds = [minute * 60 for minute in minutes]
print(seconds)
````

### 練習3（修正）― サークルの活動時間

✏️ サークルの活動時間を、1回ごとに分単位で記録したリストがあります。
1回目から順に「○回目 ○ 分」と出し、最後に「合計 ○ 分」と出すように直します。
記録の値や個数を変えても動く形にします。

考えどころ: リストの内容を自分で変え、出るはずの行を先に考えてから実行します。
第6回の「合計を入れる変数」と、第7回の「最後の番号と個数」が手がかりです。

````python
# 活動時間を順に出し、最後に合計を出す
activity = [18, 26, 22, 30, 24]
total = 1
for i in range(len(activity) + 1):
    total = activity[i]
    print(f'{i}回目 {activity[i]} 分')
print(f'合計 {total} 分')
````

### 練習4（修正）― 学食で待った時間の平均

✏️ 記録する回数を正の整数で受け取ります。その回数ぶん、学食で待った時間（分）を0以上の整数で受け取ります。
記録した回数と平均を出すように直します。

考えどころ: エラーは出ません。試す回数と待ち時間を自分で選び、出るはずの記録数と平均を先に求めてから比べます。
第7回の「合計と平均」が手がかりです。

````python
# 学食で待った時間の記録数と平均を出す
n = int(input('記録する回数: '))
waits = []
for i in range(n):
    minute = int(input('学食で待った時間（分）: '))
waits.append(minute)
average = sum(waits) / 4
print(f'記録 {len(waits)} 回')
print(f'平均 {average} 分')
````
