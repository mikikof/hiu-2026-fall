## 1. for と range

> **対応スライド**: 「for 文の形」「range の早見」

`for` の下で字下げした行を、`range` から取り出した数ごとに実行します。
`range(3)` では、0から始まる数を3つ使います。

🔹 写真のファイル名を出します。

````python
for photo in range(3):
    print(f'photo_{photo}.png')
````

`range(始め, 終わり, 刻み)` と書くと、数を増やす幅も指定できます。終わりの数は含みません。

🔹 2から3ずつ増える数を出します。

````python
for mark in range(2, 9, 3):
    print(mark)
````

## 2. 足し上げる

> **対応スライド**: 「合計を入れる変数」

横幅42 cmのポスターを3枚、隙間なく横に並べます。
合計を入れる変数を0から始め、繰り返すたびに幅を足します。

🔹 並べた幅を出します。

````python
haba = 0
for poster in range(3):
    haba = haba + 42
print(f'横幅 {haba} cm')
````

## 3. 途中を見る

> **対応スライド**: 「途中の値の確かめ方」

計算が合わないときは、小さい数で手計算した結果と比べます。
繰り返しの中に `print` を足すと、どの回から値が違うかを調べられます。確かめたら、その `print` を外します。

🔹 ポスターを並べる途中の幅も出します。

````python
haba = 0
for poster in range(3):
    haba = haba + 42
    print(poster, haba)
print(f'横幅 {haba} cm')
````

## 4. while

> **対応スライド**: 「while 文の継続条件」

`while 条件:` は、条件が True のあいだ字下げした行を繰り返します。
次のセルでは残りの秒数を減らすので、条件が False になって止まります。

🔹 11秒から4秒ずつ進めたときの残りを出します。

````python
nokori = 11
while nokori > 0:
    print(f'残り {nokori} 秒')
    nokori = nokori - 4
print('終了しました')
````

## 5. break

> **対応スライド**: 「0を打ったときの終了」「for と while の使い分け」

`break` を実行すると、繰り返しを途中で抜けます。
次のセルは合言葉を3回まで受け取り、一致したらその時点で抜けます。

🔹 `くもり`、`あおぞら` の順に打ちます。

````python
kaisu = 0
while kaisu < 3:
    kotoba = input('合言葉: ')
    kaisu = kaisu + 1
    if kotoba == 'あおぞら':
        print('一致しました')
        break
print('確認を終えました')
````

### 練習3（修正）― 倍になるゲームの得点

✏️ ゲームで得られる点は、1回目が5点で、次の回は前の回の2倍です。
遊ぶ回数を正の整数で受け取り、その回数ぶんの合計点を出すように直します。

考えどころ: 試す回数を自分で選び、仕様から合計点を求めて比べます。スライド「合計を入れる変数」と「途中の値の確かめ方」が手がかりです。

````python
# 遊んだ回数ぶんの合計点を出す
n = int(input('遊ぶ回数: '))
point = 5
score = 1
for turn in range(n - 1):
    score = point
    point = point + 2
print(f'合計 {score} 点')
````

### 練習4（修正）― 音声ガイドの案内時刻

✏️ 音声ガイドの長さを0以上の整数で受け取ります。
開始時の0秒から15秒ごとに、ガイドの長さ以下の案内時刻を出すように直します。最後に「案内を終えました」と出します。

考えどころ: 試す長さを自分で選び、出るはずの時刻を先に書き出して比べます。スライド「while 文の継続条件」が手がかりです。

````python
# 開始時から15秒ごとの案内時刻を出す
nagasa = int(input('ガイドの長さ（秒）: '))
byou = 15
while byou < nagasa:
    print(f'{byou} 秒')
    byou = byou + 15
print('案内を終えました')
````
