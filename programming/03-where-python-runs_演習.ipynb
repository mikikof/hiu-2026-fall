## 6. 変数・型・算術演算子

> **対応スライド**: 「変数に値を入れる」

`jikan = 90` と書くと、あとの行から `jikan` という名前で `90` を使えるようになります。この名前を**変数**と呼びます。
`=` は「等しい」ではありません。右の値に、左の名前を付けます。これを**代入**と呼びます。
`jikan = 90` の1行だけでは何も出ません（代入の行は値にならないためです）。

🔹 実行して、`jikan` に入れた値が出ることを確かめてください。2行目は、名前を引用符で囲んだ形です。

````python
jikan = 90
print(jikan)
print('jikan')
````

`print('jikan')` は、値の `90` ではなく、`jikan` という文字をそのまま出します。**名前は引用符で囲みません。**

🔹 セルの最後の行に名前だけを書くと、その値が表示されます。スライド「セルは最後の行の値を表示する」と同じ決まりです。

````python
jikan = 90
jikan
````

> **対応スライド**: 「値には型がある」

同じ 90 でも、書き方で**型**が変わります。`90` は整数（`int`）、`90.0` は小数（`float`）、`'90'` は文字列（`str`）です。
`type()` のカッコに値を入れると、型が分かります。

🔹 セルの最後の行に書くと、型の名前だけが出ます。

````python
type(90)
````

🔹 `print` で出すと、`<class 'int'>` の形になります。セルの最後の行の `int` と、指している型は同じです。

````python
print(type(90))
print(type(90.0))
print(type('90'))
````

> **対応スライド**: 「7つの算術演算子」

計算に使う記号を**算術演算子**と呼びます。次の7つです。

| 記号 | 計算 |
| --- | --- |
| `+` | 足し算 |
| `-` | 引き算 |
| `*` | かけ算 |
| `**` | べき乗（`2 ** 3` は 2 を3回かける） |
| `/` | 割り算 |
| `//` | 商（正の数なら、割って小数点以下を落とす） |
| `%` | 余り |

🔹 足し算、引き算、べき乗を実行します。

````python
print(90 + 15)
print(90 - 15)
print(2 ** 3)
````

> **対応スライド**: 「割り算は3通り」

`fun` には、90 分の授業 15 回ぶんの 1350 分が入ります。60 で割ると、演算子によって出るものが違います。

🔹 実行して、割り算の3行で出たものを見比べてください。

````python
fun = 90 * 15
print(fun)
print(fun / 60)
print(fun // 60)
print(fun % 60)
````

`/` は小数を、`//` は商を、`%` は余りを出します。`//` と `%` を組にすると、1350 分が 22 時間と 30 分に分かれます。

🔹 `/` は、割り切れるときも小数で出します。

````python
print(6 / 3)
print(type(6 / 3))
print(7 / 2)
````

> **対応スライド**: 「計算の順番」

1つの式に演算子がいくつかあるときは、左から順に計算するとは限りません。次の順に計算します。この順番を**優先順位**と呼びます。

1. カッコの中
2. `**`（べき乗）
3. `*` `/` `//` `%`（かけ算・割り算）
4. `+` `-`（足し算・引き算）

🔹 カッコのあるなしで、出るものを見比べてください。

````python
print(90 + 10 * 2)
print((90 + 10) * 2)
print(2 * 3 ** 2)
````

> **対応スライド**: 「文字列と数の足し算」

🔹 次のセルは、わざと失敗させてあります。文字列の `'90'` と数の `10` を `+` でつなごうとしています。実行して、いちばん下の行を見てください。

````python
print('90' + 10)
````

`TypeError` は、その計算に使えない型の組み合わせのときに出るエラーです。直すときは、引用符を外して数にするか、`+` の代わりにカンマで区切って並べます。

🔹 文字列どうしなら `+` でつながります。文字列と数でも、`*` なら止まらずに同じ文字列が繰り返されます。

````python
print('90' + '10')
print('-' * 10)
````
