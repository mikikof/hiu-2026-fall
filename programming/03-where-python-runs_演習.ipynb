## 1. セルは最後の行の値を表示する

> **対応スライド**: 「セルは最後の行の値を表示する」

`print` を書いていないのに出ることがあります。**セルの最後の行が値になるとき、Colab はそれを表示します。**

🔹 順に実行して、4つの違いを見てください。

````python
3 + 4
````

🔹 次は `print` で出した場合です。同じ 7 ですが、出どころが違います。

````python
print(3 + 4)
````

````python
'情報デザイン学部'
````

## 2. ファイルにして動かす

> **対応スライド**: 「セルの中身をファイルに書き出す」「!python3 で、ターミナルと同じ形で動かす」

`%%writefile ファイル名` をセルの**1行目**に書くと、そのセルの中身がファイルになります。
行のはじめに `!` を付けると、そのあとはターミナルへの命令になります。
どちらも Python の文法ではありません。

🔹 実行して、`Writing kamoku.py` と出ることを確かめてください。

````python
%%writefile kamoku.py
print('プログラミング入門')
3 + 4
````

🔹 作ったファイルを実行します。**何行出るか**を数えてください。

````python
!python3 kamoku.py
````

ファイルには2行入っているのに、出たのは1行だけです。
`3 + 4` の結果はどこにも出ていません。**ファイルには「最後の行の値を表示する」決まりがありません。**

🔹 ファイルの中身も見られます。

````python
!cat kamoku.py
````

## 3. print を書き忘れたファイル

> **対応スライド**: 「print を足すと、ファイルでも出る」

セルなら出るものが、ファイルにすると出ないことがあります。**エラーにはなりません。**

🔹 まず、`print` を書かないファイルを作ります。

````python
%%writefile wasure.py
'1限 プログラミング入門'
````

🔹 実行します。何も出ません。

````python
!python3 wasure.py
````

🔹 `print` を足して、ファイルを書き直します。2回目からは `Overwriting wasure.py`（上書き）と出ます。

````python
%%writefile wasure.py
print('1限 プログラミング入門')
````

🔹 実行します。今度は1行出ます。

````python
!python3 wasure.py
````

## 4. エラーの出る場所

> **対応スライド**: 「エラーの1行目だけが変わる」

同じ構文エラー（`SyntaxError`）でも、セルで実行したときとファイルで実行したときで、**1行目だけ**が変わります。

🔹 次のセルは、わざと失敗させてあります。実行して、1行目を見てください。

````python
print('こんにちは)
````

1行目は `File "/tmp/ipykernel_…/….py", line 1` の形です。
Colab は、セルの中身に一時的なファイルの名前を付けて動かしています。その名前が出ます。
名前は気にせず、`line` のうしろの行番号を見ます。

🔹 同じ誤りをファイルにして実行します。

````python
%%writefile kowareta.py
print('こんにちは)
````

🔹 作ったファイルを実行します。

````python
!python3 kowareta.py
````

今度は1行目が `File "/content/kowareta.py", line 1` です。`/content/` は Colab の標準の作業フォルダで、作ったファイルはここに置かれます。**エラーの種類も、指している場所も、セルのときと同じです**。読み方は変わりません。
